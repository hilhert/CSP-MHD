```python
import torch
import sys
sys.path.insert(0, '..')  # 如果 csp 在上一级目录

from csp.model import CSP_Seq2Seq
from csp.data import SymbolicArithmeticDataset, generate_arithmetic_sequence

# 1. 加载词表
dummy_dataset = SymbolicArithmeticDataset(1, generate_expression_func=generate_arithmetic_sequence)
vocab_size = dummy_dataset.vocab_size
char2idx = dummy_dataset.char2idx
idx2char = dummy_dataset.idx2char

# 2. 加载模型
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = CSP_Seq2Seq(vocab_size, hidden_dim=64, num_layers=3).to(device)
model.load_state_dict(torch.load('csp_symbolic.pt', map_location=device))
model.eval()

# 3. 测试
sos_idx = char2idx['<SOS>']
eos_idx = char2idx['<EOS>']
pad_idx = char2idx['<PAD>']

def generate(model, input_str, max_len=20, temperature=1.0):
    input_ids = [char2idx[c] for c in input_str] + [eos_idx]
    input_tensor = torch.tensor(input_ids, dtype=torch.long).unsqueeze(0).to(device)
    
    # 编码
    x = model.embedding(input_tensor)
    h = torch.tanh(model.encoder_proj(x))
    h_real, h_imag = h, torch.zeros_like(h)
    for layer in model.layers:
        h_real, h_imag = layer(h_real, h_imag)
    h_real_last, h_imag_last = h_real[:, -1, :], h_imag[:, -1, :]
    
    outputs = []
    h_real_dec, h_imag_dec = h_real_last, h_imag_last
    current_token = torch.full((1,), sos_idx, device=device, dtype=torch.long)
    
    for _ in range(max_len):
        emb = model.embedding(current_token)
        emb_hidden = torch.tanh(model.encoder_proj(emb))
        h_real_dec = h_real_dec + emb_hidden
        h_imag_dec = h_imag_dec + emb_hidden
        phase = torch.atan2(h_imag_dec, h_real_dec + 1e-8)
        h_cos, h_sin = torch.cos(phase), torch.sin(phase)
        h_phasor = torch.cat([h_cos, h_sin], dim=-1)
        emb_out = torch.tanh(model.decoder_proj(h_phasor))
        logits = model.output_proj(emb_out)  # [1, vocab_size]
        
        # ★ 采样解码（温度控制）
        if temperature > 0:
            probs = torch.softmax(logits / temperature, dim=-1)
            next_token = torch.multinomial(probs, 1).squeeze()
        else:
            # 贪心解码（temperature=0）
            next_token = torch.argmax(logits, dim=-1)
        
        outputs.append(next_token)
        if next_token == eos_idx:
            break
        current_token = next_token
    
    if not outputs:
        return []
    return torch.cat(outputs, dim=0).tolist()

examples = ["5 + 3 = ", "12 + 7 = ", "9 - 4 = ", "15 - 3 = "]
for inp in examples:
    # 低温度（0.8）——更保守，接近 argmax
    pred_ids = generate(model, inp, temperature=0.8)
    pred_str = ''.join([idx2char[i] for i in pred_ids if i not in [pad_idx, sos_idx, eos_idx]])
    print(f"{inp} -> {pred_str}")

# 高温度（1.2）——更随机，输出更多样
for inp in examples:
    pred_ids = generate(model, inp, temperature=1.2)
    pred_str = ''.join([idx2char[i] for i in pred_ids if i not in [pad_idx, sos_idx, eos_idx]])
    print(f"{inp} (T=1.2) -> {pred_str}")
```
